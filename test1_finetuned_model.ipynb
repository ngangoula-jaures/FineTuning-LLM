```python
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
import torch

model_id = "meta-llama/Llama-2-7b-hf"

# (Assure-toi de ne pas utiliser PeftModel.from_pretrained ici)
tokenizer = AutoTokenizer.from_pretrained(model_id)

# Chargement du modèle en FP16 pour que ça rentre dans la VRAM (comme pour ton fine-tuning)
base_model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto",
    torch_dtype=torch.float16
)

pipe = pipeline(
    task="text-generation",
    model=base_model,
    tokenizer=tokenizer,
)

# 4. Le test
prompt = "What is backpropagation?"

# On génère la suite (max_new_tokens limite la taille de la réponse)
result = pipe(prompt, max_new_tokens=50)

print("--- RÉSULTAT DU MODÈLE DE BASE ---")
print(result[0]['generated_text'])
```

## Constat : Évolution du Comportement (Base vs Fine-tuné)

### 1. Modèle de Base (`meta-llama/Llama-2-7b-hf`)
* **Comportement constaté :** Complétion de texte brute.
* **Analyse :** Le modèle d'origine ne cherche pas à répondre à l'instruction. Il se contente de calculer la probabilité des mots suivants, ce qui résulte souvent en la génération de questions similaires ou de suites de phrases dénuées d'intention conversationnelle.

### 2. Modèle Fine-tuné (Notre Version)
* **Comportement constaté :** Amorce du suivi d'instructions.
* **Analyse :** En raison de contraintes matérielles (limites GPU), le modèle n'a pas pu être entraîné sur l'entièreté du jeu de données. Cependant, on constate une nette différence dans la manière de formuler la réponse. Le modèle abandonne la simple complétion pour commencer à adopter un ton plus naturel et direct (*"For example, let us consider..."*), s'adressant à l'utilisateur.
